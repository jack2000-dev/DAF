```python
# Null audit
null_counts = df.isna().sum()
null_pct = (null_counts / len(df) * 100).round(1)
pd.DataFrame({'nulls': null_counts, 'null_%': null_pct}).query('nulls > 0').sort_values('nulls', ascending=False)
```

```python
# Distribution — replace column name
col = '[numeric_column]'
df[col].hist(bins=30)
plt.title(f'{col} Distribution')
plt.tight_layout()
plt.savefig(visual(f'{col}_distribution.png'), dpi=150)
plt.show()
```
